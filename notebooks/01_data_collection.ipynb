## 2. Load và Khảo Sát Dữ Liệu

```python
# Load all datasets
df_kaggle = pd.read_csv(RAW_DATA_DIR / 'stroke_kaggle.csv')
df_huggingface = pd.read_csv(RAW_DATA_DIR / 'stroke_huggingface.csv')
df_synthetic = pd.read_csv(RAW_DATA_DIR / 'stroke_synthetic.csv')

datasets = {
    'Kaggle': df_kaggle,
    'HuggingFace': df_huggingface,
    'Synthetic': df_synthetic
}

print("✅ All datasets loaded!")
```

```python
# Summary of all datasets
print("📊 TỔNG QUAN CÁC DATASET\n")
print(f"{'Dataset':<15} {'Rows':<10} {'Columns':<10} {'Stroke=1':<12} {'Stroke=0':<12} {'% Stroke':<10}")
print("=" * 70)

for name, df in datasets.items():
    stroke_1 = (df['stroke'] == 1).sum()
    stroke_0 = (df['stroke'] == 0).sum()
    pct = stroke_1 / len(df) * 100
    print(f"{name:<15} {len(df):<10} {len(df.columns):<10} {stroke_1:<12} {stroke_0:<12} {pct:.2f}%")
```
